## 3. Inspect the file before reading it

```python
from pathlib import Path

path = Path(filename)
print("Exists:", path.exists())
print("Size (bytes):", path.stat().st_size if path.exists() else "not found")
print("\\nFirst 12 lines:")
if path.exists():
    print("\\n".join(path.read_text().splitlines()[:12]))
```

Questions to ask before ingesting a file:
- Is there a header?
- What separates fields?
- Are there comment lines?
- What do missing values look like?
- Are units and date formats clear?

The extension `.txt` does not tell us the internal structure. This particular file is comma-delimited text with comment lines.

## 4. Read the observations

```python
import pandas as pd

df = pd.read_csv(path, comment='#', parse_dates=['date'])
df
```

## 5. Inspect the data structure

```python
print(df.shape)
print(df.dtypes)
print(df.info())
print(df.describe(numeric_only=True))
```

## 6. A simple plot

```python
import matplotlib.pyplot as plt

for station, group in df.groupby('station'):
    plt.plot(group['date'], group['temperature_C'], marker='o', label=station)

plt.ylabel('Temperature (°C)')
plt.xlabel('Date')
plt.title('Illustrative daily temperatures')
plt.legend()
plt.grid(alpha=0.25)
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```
